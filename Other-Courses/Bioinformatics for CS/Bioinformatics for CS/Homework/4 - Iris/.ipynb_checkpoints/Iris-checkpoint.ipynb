# Perceptron

```python
class Perceptron(object):
    # Initiating the learning rate and number of iterations
    def __init__(self, eta = 0.1, iter = 50):
        self.eta = eta
        self.iter = iter
        
    def fit(self, X, y):
        self.w_ = np.zeros(X.shape[1])
        self.errors_ = []

        for _ in range(self.iter):
            error = 0
            for xi, target in zip(X, y):
                update      = self.eta * (target - self.predict(xi))
                self.w_[:] += update*xi
                error      += int(update != 0.0)
            self.errors_.append(error)
        return self

    def predict(self, x):
        return np.where(np.dot(x, self.w_[:]) >= 0.0, 1, -1)
```

```python
def main():
    dataFile = 'iris.txt'

    X, y = formatIris(dataFile)   # X includes 1 for the bias: X = [[1,x11, x12], [1,x21,x22], .... [1,xN1, xN2]]T

    eta = 0.1   # learning rate
    
    scaler = StandardScaler()
    X = scaler.fit_transform(X)
    
    X_std = np.copy(X)
    X_std[:,1] = (X[:,1] - X[:,1].mean()) / X[:,1].std()   # sepal len
    X_std[:,2] = (X[:,2] - X[:,2].mean()) / X[:,2].std()   # petal len

    # Perceptron
    ppn = Perceptron(eta)   #, iter=len(y))
    ppn.fit(X,y)
    
    plot_regions(X, y, classifier=ppn)
```

```python
def formatIris(dataset):
    X = []
    y = []
    with open(dataset, 'r') as f:
        for line in f:
            sline = line.replace('\n','')
            t = sline.split(',')
            if   t[4]=='Iris-versicolor':   val = 1
            elif t[4]=='Iris-virginica':    val = -1
            else: val = 0

            X.append([1, float(t[0]), float(t[2])])   # [1, sepal len, petal len]
            y.append(val)
    return np.array(X), np.array(y)
```

```python
def plot_regions(X, y, classifier, resolution = 0.02):
    # setup marker generator and color map
    plt.figure(figsize=(10,10))
    markers = ('o', 'x', 's', '^', 'v')
    colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
    cmap = ListedColormap(colors[:len(np.unique(y))])

    # plot the decision surface
    x1_min, x1_max = X[:,1].min() - 1, X[:,1].max()+1
    x2_min, x2_max = X[:,2].min() - 1, X[:,2].max()+1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution), np.arange(x2_min, x2_max, resolution))
   
    coordinates = np.array([np.ones(len(xx1.ravel())), xx1.ravel(), xx2.ravel()])
    region = np.dot(coordinates.T, classifier.w_[:])
    Z = np.where(region >= 0.0, 1, -1)
    Z = Z.reshape(xx1.shape)
    plt.contourf(xx1, xx2, Z, alpha=0.4, cmap=cmap)
   
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())
    for idx, cl in enumerate(np.unique(y)):
        plt.scatter(x = X[y==cl, 1], y = X[y==cl, 2], alpha=0.8, color=cmap(idx), marker=markers[idx], label=cl)

    plt.title("Perceptron")
    plt.xlabel('sepal length')
    plt.ylabel('petal length')
    plt.legend(loc='upper right')
    plt.show()
```

```python
main()
```

# Linear Regression Algorithm

```python
dataset = dataset.replace( {'Iris-versicolor' : -1, 'Iris-virginica' : 1})
dataset
```

```python
dataset = dataset.drop(['sepal_width','petal_width'], axis=1)
```

```python
dataset.head()
```

```python
dataset.tail()
```

```python
x = dataset.iloc[:,[0,1]]
y = dataset['Species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
classifier = LinearRegression()
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)

print('Accuracy of logistic regression classifier on test set: {:.2f}'.format(classifier.score(X_test, y_test)))
```

```python
plt.figure(figsize=(10,10))
markers = ('o', 'x', 's', '^', 'v')
colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
cmap = ListedColormap(colors[:len(np.unique(y))])

X_set, y_set = X_train, y_train

X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))

plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.4, cmap = cmap)

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], color=cmap(i),
                marker=markers[i], label = j)
    
plt.title('Linear Regression')
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.legend(loc='upper right')
plt.show()
```

# Logistic Regression Algorithm

```python
x = dataset.iloc[:,[0,1]]
y = dataset['Species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)

print('Accuracy of logistic regression classifier on test set: {:.2f}'.format(classifier.score(X_test, y_test)))
```

```python
plt.figure(figsize=(10,10))
markers = ('o', 'x', 's', '^', 'v')
colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
cmap = ListedColormap(colors[:len(np.unique(y))])

X_set, y_set = X_train, y_train

X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))

plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.4, cmap = cmap)

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], color=cmap(i),
                marker=markers[i], label = j)
    
plt.title('Logistic Regression')
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.legend(loc='upper right')
plt.show()
```
