# Step 1: Normalizing Data

```python
matrixNorm = np.copy(data_matrix)
for i in range(numCol):
    matrixNorm[:,i] = (data_matrix[:,i]-data_matrix[:,i].mean())/data_matrix[:,i].std()
```

```python
norm = {'Volume':matrixNorm[:,0], 'Bulk':matrixNorm[:,1], 'Polarity':matrixNorm[:,2], 'pI':matrixNorm[:,3], 
        'Hyd. 1':matrixNorm[:,4], 'Hyd. 2':matrixNorm[:,5], 'Surface_Area':matrixNorm[:,6], 'Fractal_Area':matrixNorm[:,7]}
```

```python
Normalized = pd.DataFrame(data=norm)
Normalized
```

# Step 2: Covariance Matrix

```python
covMatO = np.zeros(numCol*numCol)
covMat = np.reshape(covMatO, (numCol, numCol))
for row in range(numCol):
    for col in range(numCol):
        covMat[row, col] = (matrixNorm[:,row].T).dot(matrixNorm[:,col])/(numRow)
```

```python
cov = {'Volume':covMat[:,0], 'Bulk':covMat[:,1], 'Polarity':covMat[:,2], 'pI':covMat[:,3], 
       'Hyd. 1':covMat[:,4], 'Hyd. 2':covMat[:,5], 'Surface_Area':covMat[:,6], 'Fractal_Area':covMat[:,7]}
```

```python
cov_matrix = pd.DataFrame(data=cov)
cov_matrix
```

```python
f = plt.figure(figsize=(17, 13))
corrMatrix = cov_matrix.corr()
sn.heatmap(corrMatrix, annot=True)
corrMatrix.style.background_gradient(cmap='coolwarm')
plt.show()
```

# Step 3: Eigenvalues and Eigenvectors

```python
eigen_values, eigen_vectors = np.linalg.eig(cov_matrix)
```
