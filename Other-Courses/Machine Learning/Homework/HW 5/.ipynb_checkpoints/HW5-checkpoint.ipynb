P1. Describe two techniques to select the right number of clusters when using K-Means.

Two possible techniques to selecting the right number of clusters when using K-Means is the "Elbow" Method and The Silhouette Method.

The Elbow method uses the sum of squares at each number of clusters is calculated ad graphed, and the user looks for a change of slope from steep to shallow (an elbow) to determine the number of clusters. This method is inexact, but still helpful.

The Silhouette Method uses a Silhouette coefficient for each instance and cab be computed as     (b - a)/max(a, b); where a is the mean distance to the other distances in the same cluster (mean intra-cluster distance); and b is the mean of the nearest-cluster distance, where a value of +1 is the instance is well inside its own cluster; 0, the instance is close to a cluster boundary; and -1, where the instance is assigned to the wrong cluster

P2. What is the difference between Hard clustering and soft clustering?

Hard Clustering: Assigning each instance to a single cluster

Soft Clustering: Assigning each instance a score per cluster (like distance between the instance and the centroid or a similarity score)

P4. Cluster the following points, (18, 10), (21, 11), (22, 22), (24, 15), (26, 12), (26, 13), (27, 14), (30, 33), (31, 39), (35, 37), (39, 44), (40, 27), (41, 29), (42, 20), (44, 28), (46, 21), (47, 30), (48, 31), (49, 23), (54, 24) use the numbers as pairs of x and y values which represent their locations. All distances are measured with Euclidean distance.

```python
X = np.array([[18, 10], [21, 11], [22, 22], [24, 15], [26, 12], 
              [26, 13], [27, 14], [30, 33], [31, 39], [35, 37], 
              [39, 44], [40, 27], [41, 29], [42, 20], [44, 28], 
              [46, 21], [47, 30], [48, 31], [49, 23], [54, 24]])

print("Array dimension:", X.ndim)
print("Array shape:", X.shape)
print("X =", X[:, 0])
print("Y =", X[:, 1])
```

```python
plt.rcParams['figure.figsize'] = (7, 7)
plt.xlabel('X', fontsize = 20)
plt.ylabel('Y', fontsize = 20);
plt.grid(True)
plt.axis([15, 55, 5, 45])

plt.scatter(X[:, 0], X[:, 1], s = 50)
plt.show()
```

P4.1 Train the Kmeans clustering algorithm on the data with (24, 15), (30, 33), and (54, 24) as the initial cluster centers. Predict the cluster of each data point and show the result. 

```python
kmeans = KMeans(n_clusters = 2).fit(X)
pred_y = kmeans.predict([[24, 25], [30, 33], [54, 24]])

labels = kmeans.labels_
centroids = kmeans.cluster_centers_
c = Counter(labels)

pred_y
```

```python
print ("centroids : ")
print (centroids)
print("\n")
print ("labels : ")
print (labels)
```

P4.2 Visualise the clusters 

```python
plt.xlabel('X', fontsize = 20)
plt.ylabel('Y', fontsize = 20);
plt.grid(True)
plt.axis([15, 55, 5, 45])

plt.scatter([X[:, 0]], [X[:, 1]], c = kmeans.labels_.astype(int), s = 50)
plt.scatter(centroids[:, 0], centroids[:, 1], c = 'b', s = 150, alpha = 0.7);

plt.show()
```

P5. Programming assignment (KMeans clustering)

```python
# Import the dataset
dataset = pd.read_csv('Mall_Customers.csv')
X = dataset.iloc[:, [3, 4]].values
print(dataset.head())
```

```python
# Plotting the values to understand the spread
plt.scatter(X[:, 0], X[:, 1], c = 'b', s = 50)
plt.grid(True)
```

```python
# P5. 1 Use the elbow method to find the optimal number of clusters through a loop (1 to 10 clusters) 
# and visualize the result on a 2D plot
# Set init = 'k-means++'

plt.rcParams['figure.figsize'] = (8, 8)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.grid(True)

wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss)

plt.show()
```

```python
# P5. 2 Fit K-Means to the dataset and use the optimal number of clusters that you 
# found in the previous part and show the result

kmeans = KMeans(n_clusters = 5, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
pred_y = kmeans.fit_predict(X)
centers = kmeans.cluster_centers_
centers
```

```python
# P5. 3 Visualize the clusters (# xlabel('Annual Income (k$)' and ylabel('Spending Score (1-100)'))

# Use different colors for clusters and label them

#plt.scatter(X[:,0], X[:,1], c = kmeans.labels_.astype(float))
plt.scatter(X[pred_y == 0, 0], X[pred_y == 0, 1], s = 50, c = 'purple',  label = 'Cluster 1')
plt.scatter(X[pred_y == 1, 0], X[pred_y == 1, 1], s = 50, c = 'blue',    label = 'Cluster 2')
plt.scatter(X[pred_y == 2, 0], X[pred_y == 2, 1], s = 50, c = 'green',   label = 'Cluster 3')
plt.scatter(X[pred_y == 3, 0], X[pred_y == 3, 1], s = 50, c = 'orange',  label = 'Cluster 4')
plt.scatter(X[pred_y == 4, 0], X[pred_y == 4, 1], s = 50, c = 'magenta', label = 'Cluster 5')


plt.scatter(centers[:, 0], centers[:, 1], s = 100, marker = 's', c = 'red', label = 'Centroids')
plt.xlabel('Annual Income (k$)', fontsize = 20)
plt.ylabel('Spending Score (1-100)', fontsize = 20) 
plt.text(67, 49, "cluster 1", fontsize = 15, c = 'red')
plt.text(29, 77, "cluster 2", fontsize = 15, c = 'red')
plt.text(90, 80, "cluster 3", fontsize = 15, c = 'red')
plt.text(30, 20, "cluster 4", fontsize = 15, c = 'red')
plt.text(90, 19, "cluster 5", fontsize = 15, c = 'red')
plt.grid(True)
plt.legend()
plt.show()
```
