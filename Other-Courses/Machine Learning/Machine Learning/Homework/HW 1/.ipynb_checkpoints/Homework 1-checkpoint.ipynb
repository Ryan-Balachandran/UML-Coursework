Question 3: What is the difference between a model parameter and a learning algorithm's hyperparameter?

Answer 3: A model parameter helps the algorithm predict what to do with a new instance (usually with more than one parameter) while a hyperparameter is a parameter for the learning algorithm and is given before the algorithm is trained.

Question 4: What is a test set, and why would you want to use it?

Answer 4: A test set is a set of data you use to test your model to see how it performs. A test set is necessary to assess the performance of the module.

---------

Question 5: What is the purpose of a validation set?

Answer 5: The validation set is used to compare different training models to find the best model to solve a given problem.

---------

Question 6: What can go wrong if you tune hyperparameters using the test set?

Answer 6: You would not want to tune the hyperparameters to the test set because the model would be trained only to that specific set and would not perform well on other data sets.

Task: create a normalized form of the following data set (accessed through UCI Machine Learning Repository) whose values range exactly between 0 and 1 so that the minimum has value 0 and maximum has value 1.

```python
# Load the data 
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
data = np.genfromtxt(url, delimiter = ',', dtype = 'float', usecols = [0])
```

```python
# Find the maximum and minimum values of data and print them: Smax and Smin
Smax, Smin = max(data), min(data)

# Solution: Smax, Smin = np.max(data), np.min(data)
print(Smax, Smin)
```

After writing your own code, we like to confirm that built-in functions in Scikit-Learn provide identical results. There are at least two techniques for normalizing the data using *sklearn.preprocessing*.

```python
# Use (x - Smin)/(Smax - Smin) to normalize the data 
data_preprocessed = (data - Smin)/(Smax - Smin)
print(data_preprocessed)
```

After writing your own code, we like to confirm that built-in functions in Scikit-Learn provide identical results. There are at least two techniques for normalizing the data using *sklearn.preprocessing*.

```python
# Technique 1: using minmax_scale 
from sklearn.preprocessing import minmax_scale

data_preprocessed2 = minmax_scale(data)
print(data_preprocessed2)
```

```python
# Technique 2: using the``Transformer`` API MinMaxScaler 
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data_preprocessed3 = scaler.fit_transform(data)
```
